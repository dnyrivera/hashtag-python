#### 12. Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês.


```python
value_hour = float(input("Enter the value by hour: "))
hours_worked = float(input("Enter how many hours worked: "))
```

##### Calcule o salário bruto (horas \* salario por hora)


```python
gross_pay: float = value_hour * hours_worked
print(f"Your total salary is ${gross_pay} ")
```

##### Calcule o desconto do IR (11% do salário bruto)


```python
irpf_discount: float = gross_pay * 0.11
print(f"IRPF Discount: ${irpf_discount}")
```

##### Calcule o desconto do INSS (8% do salário bruto)


```python
inss_discount: float = gross_pay * 0.08
print(f"INSS Discount: ${inss_discount}")
```

##### Calcule o desconto do sindicato (5% do salário bruto)


```python
sindicate_discount: float = gross_pay * 0.05
print(f"Sindicate Discount: ${sindicate_discount}")
```

##### Calcule o salário líquido (salário bruto - descontos)


```python
net_pay = gross_pay - (irpf_discount + inss_discount + sindicate_discount)
print(f"Your Net pay is : ${net_pay}")
```

#### 14. Faça um programa que peça o tamanho de um arquivo para download (em MB) e a velocidade de um link de Internet (em Mbps), calcule e informe o tempo aproximado de download do arquivo usando este link (em minutos).

Detalhe: MB significa megabyte, Mb (com b minúsculo) significa megabit. Um megabit é 1/8 de um megabyte.


```python
file_size = float(input("File Size in MB: "))
internet_connection = float(input("Internet Speed in Mbps: "))
download_time = ((file_size * 8) / internet_connection) / 60
print(f"The total time to download the file is {download_time:.2f} minutes.")
```
