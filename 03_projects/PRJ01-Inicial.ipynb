### Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
import time
from datetime import datetime

import pandas as pd
import pyautogui
```

```python
# Passo 01 : Entrar no sistema da empresa (link do drive)
pyautogui.PAUSE = 0.7  # Pausa de 0.8s entre cada comando
link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga"

# Passo 01: Entrar no sistema da empresa (link do drive)
# 1. Abrir o Spotlight e digitar Chrome
pyautogui.hotkey("command", "space")
pyautogui.write("Google Chrome")
pyautogui.press("enter")

time.sleep(2)

# 2. Abrir uma nova aba para garantir que o cursor está na barra de endereço
pyautogui.hotkey("command", "t")
pyautogui.write(link)
pyautogui.press("enter")
time.sleep(2)
```

```python
# Step 2: Navigate to the dataset folder
time.sleep(2)
pyautogui.click(x=367, y=397)  # 1st click: focuses the Chrome window
pyautogui.click(x=367, y=397)  # 2nd click: selects the folder in Google Drive
pyautogui.press("enter")  # Opens the selected folder
time.sleep(2)  # Wait for Google Drive to load the file inside

# Step 3: Export the database (download the file)
pyautogui.click(x=367, y=397)
pyautogui.rightClick(x=367, y=397)  # Same as click(..., button="right")
pyautogui.click(x=423, y=462)  # Left-click "Download" in the menu
time.sleep(2)  # Wait for the download to finish

# Step 4: Type the file name and save in the folder download
pyautogui.click(x=1010, y=271)  # Save As box

# Prepare the file text
today = datetime.now().strftime("%d-%m-%Y_%H-%M")
pyautogui.write("DEZ-SALES_" + today)  # -> "17-11-2026_02-20")

pyautogui.click(x=1695, y=1223)
time.sleep(2)
```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
# Step 4 : Calculate the measures (Sales / Products Quantity)

# Open file in the path
file_path = r"/Users/rivera/Downloads/DEZ-SALES_17-11-2026_02-33.xlsx"
df_sales = pd.read_excel(file_path)

# Chech the dataframe
display(df_sales)

# Sum the Sales Amount
sales_amount = df_sales["Valor Final"].sum()

# Sum Qty Products
products_amount = df_sales["Quantidade"].sum()

print(f"Sales Amount: {sales_amount} | Products Quantity: {products_amount}")
```
