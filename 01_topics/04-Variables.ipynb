## 1. Numeric Types

Python has integers (`int`), decimal numbers (`float`), and complex numbers (`complex`).


```python
age = 25                 # int: a whole number
temperature = 21.5       # float: a decimal number
complex_number = 2 + 3j  # complex: real and imaginary parts

print(age, type(age))
print(temperature, type(temperature))
print(complex_number, type(complex_number))
```

<a id="text-type"></a>

## 2. Text Type

A string (`str`) stores text inside single or double quotes.


```python
course_name = "Python Basics"
first_letter = course_name[0]

print(course_name, type(course_name))
print(f"The first letter is {first_letter}.")
```

<a id="boolean-type"></a>

## 3. Boolean Type

A Boolean (`bool`) is either `True` or `False`. It is commonly used in conditions.


```python
is_learning = True
is_finished = False
is_adult = age >= 18

print(is_learning, type(is_learning))
print("Is the student an adult?", is_adult)
```

<a id="none-type"></a>

## 4. None Type

`None` represents the absence of a value. Its type is `NoneType`.


```python
result = None

print(result, type(result))
print(result is None)
```

<a id="sequence-types"></a>

## 5. Sequence Types

Sequences keep items in order. A `list` is changeable, a `tuple` is immutable, and a `range` represents a sequence of integers.


```python
languages = ["Python", "JavaScript", "SQL"]  # list
coordinates = (40.4168, -3.7038)              # tuple
numbers = range(1, 6)                         # range

languages.append("Java")
print(languages, type(languages))
print(coordinates, type(coordinates))
print(list(numbers), type(numbers))
```

<a id="set-types"></a>

## 6. Set Types

A `set` stores unique items and is changeable. A `frozenset` is an immutable set.


```python
unique_numbers = {1, 2, 2, 3}
fixed_permissions = frozenset({"read", "write"})

print(unique_numbers, type(unique_numbers))
print(fixed_permissions, type(fixed_permissions))
```

<a id="mapping-type"></a>

## 7. Mapping Type

A dictionary (`dict`) stores values as key-value pairs.


```python
student = {
    "name": "Alex",
    "age": 25,
    "is_learning": True,
}

print(student, type(student))
print("Student name:", student["name"])
```

<a id="binary-types"></a>

## 8. Binary Types

Binary types store byte data. `bytes` is immutable, `bytearray` is changeable, and `memoryview` accesses binary data without copying it.


```python
encoded_text = b"Python"
editable_bytes = bytearray([65, 66, 67])
byte_view = memoryview(encoded_text)

print(encoded_text, type(encoded_text))
print(editable_bytes, type(editable_bytes))
print(byte_view, type(byte_view))
```

<a id="checking-and-changing-types"></a>

## 9. Checking and Changing Types

Use `type()` to inspect a value, `isinstance()` to test its type, and constructors such as `int()` or `str()` to convert it.


```python
price_text = "19"
price_number = int(price_text)

print(type(price_text))
print(type(price_number))
print(isinstance(price_number, int))
print(float(price_number), str(price_number))
```

<a id="dynamic-typing"></a>

## 10. Dynamic Typing

A Python variable can refer to values of different types at different times. The value has a type; the variable name does not have a permanently fixed type.


```python
value = 10
print(value, type(value))

value = "ten"
print(value, type(value))
```
