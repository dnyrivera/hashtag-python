## 1. Basic Syntax and Indentation

```python
if condition:
    # This block runs when the condition is true.
    print("The condition is true")
```

**Rules to remember**

- Write the keywords in lowercase: `if`, `elif`, `else`.
- End each `if`, `elif`, and `else` header with a colon (`:`).
- Indent the body consistently. Use **4 spaces** per level; do not mix tabs and spaces.
- Align `elif` and `else` with the `if` they belong to.
- `if` and `elif` need a condition; `else` has **no condition**.
- A block needs at least one statement. Use `pass` if you intentionally need an empty placeholder.
- Parentheses around a condition are optional. Use them when they make grouping clearer.
- A line that returns to the previous indentation level is outside the block.

A comparison such as `10 > 5` produces a Boolean value: `True` or `False`.


````python
is_learning = True

if is_learning:
    print("Read the rule.")
    print("Run the example.")

print("This line is outside the if block.")
````

**Expected output:**

```text
Read the rule.
Run the example.
This line is outside the if block.
```

**Try it:** Set `is_learning = False`. Only the last message will print.

## 2. Comparison Operators

Use comparisons to ask questions about values.

| Operator | Meaning                  | Example  | Result  |
| -------- | ------------------------ | -------- | ------- |
| `==`     | Equal to                 | `5 == 5` | `True`  |
| `!=`     | Not equal to             | `5 != 5` | `False` |
| `>`      | Greater than             | `8 > 5`  | `True`  |
| `<`      | Less than                | `8 < 5`  | `False` |
| `>=`     | Greater than or equal to | `5 >= 5` | `True`  |
| `<=`     | Less than or equal to    | `5 <= 5` | `True`  |

**Boundary rule:** `>` excludes equality; `>=` includes it. The same distinction applies to `<` and `<=`.

**Assignment versus comparison:** `score = 80` stores a value; `score == 80` compares values.


````python
score = 80

print("Equal to 80:", score == 80)
print("Different from 80:", score != 80)
print("Above 80:", score > 80)
print("At least 80:", score >= 80)
print("Below 80:", score < 80)
print("At most 80:", score <= 80)
````

**Expected output:**

```text
Equal to 80: True
Different from 80: False
Above 80: False
At least 80: True
Below 80: False
At most 80: True
```

**Try it:** Try `score = 79`, then `score = 81`. Predict all six results first.

## 3. `if`: Do Something When True

Use a single `if` when you only need an action for the true case. If the condition is false, Python skips its block.

**Example: sales target** (invented data). A product meets its target when sales are **equal to or greater than** 50,000 units, so use `>=`.
Underscores in numbers improve readability: `50_000` means the same as `50000`.


````python
sales_target = 50_000
units_sold = 65_300

if units_sold >= sales_target:
    print(f"Target reached! Sold {units_sold:,} units.")

print("Sales check complete.")
````

**Expected output:**

```text
Target reached! Sold 65,300 units.
Sales check complete.
```

**Try it:** Test `units_sold` with `49_999`, `50_000`, and `50_001`. The completion message always prints.

## 4. `if / else`: Choose Between Two Paths

Use `else` for what should happen when the `if` condition is false. **Exactly one of the two blocks runs.**

```python
if condition:
    # True case
    ...
else:
    # False case
    ...
```

The dots above are placeholders; replace them with your actions.


````python
sales_target = 50_000
units_sold = 15_000

if units_sold >= sales_target:
    print(f"Target reached! Sold {units_sold:,} units.")
else:
    units_missing = sales_target - units_sold
    print(f"Target not reached. Need {units_missing:,} more units.")
````

**Expected output:**

```text
Target not reached. Need 35,000 more units.
```

**Try it:** Set `units_sold = 50_000`. Equality belongs to the success branch.


### Another example: even or odd

The remainder operator `%` helps check divisibility. `number % 2 == 0` means the number has no remainder when divided by 2, so it is even.


````python
number = 7

if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")
````

**Expected output:**

```text
7 is odd.
```

**Try it:** Try `8`, `0`, and `-3`. Zero is even.

## 5. `if / elif / else`: Choose Among Several Paths

`elif` means **else if**. Python checks conditions **from top to bottom** and runs only the **first matching block**.
Later conditions in that chain are skipped. If none match, `else` runs, if present.

**Structure:** one `if`, zero or more `elif` branches, and optionally one final `else`.

**Example: a practice grading system**

- Outside 0–100: invalid score.
- 90–100: A.
- 80–less than 90: B.
- 70–less than 80: C.
- Below 70: needs practice.

Check the invalid range first, then check grade thresholds from highest to lowest.


````python
score = 85

if score < 0 or score > 100:
    print("Invalid score: use a value from 0 to 100.")
elif score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
else:
    print("Needs practice")
````

**Expected output:**

```text
Grade B
```

**Try it:** Test `-1`, `0`, `69`, `70`, `79`, `80`, `89`, `90`, `100`, and `101`. These values check the boundaries. Section 7 explains `or`.

## 6. Separate `if` Statements Versus an `elif` Chain

| Structure                    | What happens                                                | When to use it                  |
| ---------------------------- | ----------------------------------------------------------- | ------------------------------- |
| Separate `if` statements     | Every condition is checked; several blocks can run.         | Independent actions or checks.  |
| One `if / elif / else` chain | Only the first matching block runs, or the fallback `else`. | Choose one category or outcome. |

For a score of 95, both `score >= 70` and `score >= 90` are true.


````python
score = 95

print("Independent checks:")
if score >= 70:
    print("Passed")
if score >= 90:
    print("Excellent result")

print("One category:")
if score >= 90:
    print("Excellent result")
elif score >= 70:
    print("Passed")
else:
    print("Needs practice")
````

**Expected output:**

```text
Independent checks:
Passed
Excellent result
One category:
Excellent result
```

**Try it:** In the category chain, put `score >= 70` first and `score >= 90` second. Notice why the order changes the result.

## 7. Combine Conditions: `and`, `or`, `not`

| Operator | Rule                                                 | Example                   |
| -------- | ---------------------------------------------------- | ------------------------- |
| `and`    | Both conditions must be true.                        | `has_ticket and has_id`   |
| `or`     | At least one condition must be true, including both. | `is_member or has_coupon` |
| `not`    | Reverses the truth value.                            | `not is_closed`           |

### Boolean truth table

| A       | B       | A `and` B | A `or` B |
| ------- | ------- | --------- | -------- |
| `True`  | `True`  | `True`    | `True`   |
| `True`  | `False` | `False`   | `True`   |
| `False` | `True`  | `False`   | `True`   |
| `False` | `False` | `False`   | `False`  |

`not True` is `False`; `not False` is `True`.

For these operators, Python applies `not` before `and`, and `and` before `or`. Use parentheses to make your intent obvious.


````python
has_ticket = True
has_id = False
is_member = False
has_coupon = True
is_closed = False

if has_ticket and has_id:
    print("Entry approved.")
else:
    print("Bring both your ticket and ID.")

if is_member or has_coupon:
    print("Discount available.")

if not is_closed:
    print("The shop is open.")
````

**Expected output:**

```text
Bring both your ticket and ID.
Discount available.
The shop is open.
```

**Try it:** Set `has_id = True`, then try all four combinations of `is_member` and `has_coupon`.


### Ranges and grouping

Python supports chained comparisons: `18 <= age <= 25` means `age >= 18 and age <= 25`.
Both endpoints are included here.


````python
age = 22
is_student = True
has_special_pass = False

if (18 <= age <= 25 and is_student) or has_special_pass:
    print("Eligible for the study event.")
else:
    print("Not eligible for the study event.")
````

**Expected output:**

```text
Eligible for the study event.
```

**Try it:** Try age `26` without a special pass, then with a special pass. The pass is an alternative route to eligibility.


### Short-circuit evaluation

Python checks from left to right. With `and`, it stops when an operand is falsy. With `or`, it stops when an operand is truthy.
This can protect an operation that is only valid after an earlier check.


````python
items = 0
total_cost = 100

if items > 0 and total_cost / items < 30:
    print("Average cost is below 30.")
else:
    print("No items, or average cost is at least 30.")
````

**Expected output:**

```text
No items, or average cost is at least 30.
```

**Try it:** Try `items = 5`. With zero items, Python skips the division, avoiding division by zero.

## 8. Truthy and Falsy Values, Membership, and `None`

### Truthy and falsy values

An `if` can check a value directly. Python interprets that value as true or false.

| Common falsy values       | Meaning          |
| ------------------------- | ---------------- |
| `False`                   | Boolean false    |
| `None`                    | No value         |
| `0`, `0.0`                | Numeric zero     |
| `""`                      | Empty string     |
| `[]`, `()`, `{}`, `set()` | Empty containers |

Nonzero numbers and nonempty built-in strings and containers are truthy.
**Watch out:** `"False"`, `"0"`, and `" "` are nonempty strings, so they are truthy.
Use `bool(value)` to see a value's truth value.

`and` and `or` return an operand, which is not always a Boolean: `"" or "Guest"` returns `"Guest"`. `not` always returns a Boolean.


````python
shopping_cart = []

if shopping_cart:
    print("Ready to check out.")
else:
    print("Your cart is empty.")

print("Empty text:", bool(""))
print("Text saying False:", bool("False"))
print("Negative number:", bool(-1))
````

**Expected output:**

```text
Your cart is empty.
Empty text: False
Text saying False: True
Negative number: True
```

**Try it:** Set `shopping_cart = ["notebook"]` and run the cell again.


### Membership: `in` and `not in`

Use `in` to check whether an item belongs to a collection, and `not in` to check that it does not.
For strings, `in` checks for a substring. For dictionaries, it checks keys.
String comparisons and membership checks are case-sensitive.


````python
role = "editor"
allowed_roles = ["admin", "editor"]

if role in allowed_roles:
    print("You can edit this document.")
else:
    print("View-only access.")

if "guest" not in allowed_roles:
    print("Guests cannot edit.")
````

**Expected output:**

```text
You can edit this document.
Guests cannot edit.
```

**Try it:** Try `role = "guest"`, then `role = "Editor"`. Capitalization matters.


### Check for a missing value with `is None`

Use `value is None` or `value is not None` when you specifically mean “missing” or “present.”
A value of zero can be valid data even though it is falsy.

Use `==` to compare values. `is` checks object identity; do not use it to compare ordinary numbers or strings.


````python
score = 0

if score is None:
    print("No score has been recorded.")
else:
    print(f"Recorded score: {score}")
````

**Expected output:**

```text
Recorded score: 0
```

**Try it:** Set `score = None`. A missing score is different from a score of zero.

## 9. Nested Conditionals

A nested conditional is an `if` inside another conditional block. Use it when a decision depends on a previous decision.
Each new level adds another 4 spaces of indentation.

**Example:** first check whether a user is logged in, then check whether that user can edit.


````python
is_logged_in = True
can_edit = False

if is_logged_in:
    if can_edit:
        print("Open the editor.")
    else:
        print("Open the read-only view.")
else:
    print("Please log in.")
````

**Expected output:**

```text
Open the read-only view.
```

**Try it:** Test all four combinations of `is_logged_in` and `can_edit`. When logged out, the inner check is skipped.

## 10. Conditional Expressions

Use a conditional expression to choose between **two simple values**:

```python
value_if_true if condition else value_if_false
```

This is often called a ternary expression. Use a normal `if / else` when branches contain several actions or the expression becomes hard to read.


````python
score = 72
result = "Pass" if score >= 60 else "Try again"
print(result)

# The same decision written as a normal if / else:
if score >= 60:
    result = "Pass"
else:
    result = "Try again"

print(result)
````

**Expected output:**

```text
Pass
Pass
```

**Try it:** Set `score = 59`, then `score = 60`. Both forms should agree.

## 11. Common Mistakes and a Quick Reference

| Mistake                                          | Better version                    | Why                                                 |
| ------------------------------------------------ | --------------------------------- | --------------------------------------------------- |
| `if score = 80:`                                 | `if score == 80:`                 | `=` assigns; `==` compares.                         |
| `if score >= 80`                                 | `if score >= 80:`                 | A header needs a colon.                             |
| Unindented body                                  | Indent the body by 4 spaces.      | Indentation defines the block.                      |
| `else score < 80:`                               | `else:` or `elif score < 80:`     | `else` never takes a condition.                     |
| Checking `>= 70` before `>= 90` in a grade chain | Check the higher threshold first. | Only the first matching branch runs.                |
| `if role == "admin" or "editor":`                | `if role in ("admin", "editor"):` | The nonempty string `"editor"` is always truthy.    |
| `if score:` to detect a recorded score           | `if score is not None:`           | Zero may be a valid score.                          |
| `if name is "Alex":`                             | `if name == "Alex":`              | Compare values with `==`.                           |
| Using `>` when equality should count             | Use `>=`.                         | Test the exact boundary.                            |
| Comparing `"18" >= 18`                           | Convert numeric text first.       | A string and an integer cannot be ordered this way. |

**Input reminder:** `input()` returns text. For valid integer text, use `age = int(age_text)` before comparing it with a number. Invalid text such as `"eighteen"` raises `ValueError`; handling that belongs to exception handling.

### Quick reference

```python
if condition:
    action()
elif other_condition:
    other_action()
else:
    fallback_action()
```

The function names above are placeholders for your own actions, such as `print("Hello")`.

- One optional action → `if`.
- Two alternative paths → `if / else`.
- One of several categories → `if / elif / else`.
- Multiple independent actions → separate `if` statements.
- All requirements → `and`; any requirement → `or`; reverse a truth value → `not`.
- Value in a range → `lower <= value <= upper`.
- Membership → `in` or `not in`; missing value → `is None`.
- Choose between two short values → conditional expression.
